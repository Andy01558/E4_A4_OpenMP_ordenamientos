```python
%%writefile practica3_2_final.cu
#include <cstdio>
#include <cstdlib>
#include <cuda_runtime.h>

__global__ void sumarArreglos(int* A, int* B, int* C, int N, bool verbose) {
    int globalIdx = blockIdx.x * blockDim.x + threadIdx.x;
    if (globalIdx < N) {
        C[globalIdx] = A[globalIdx] + B[globalIdx];
        if (verbose && globalIdx % 5 == 0) {
            printf("[DEVICE] [Bloque: %d] [Hilo: %d] [Indice Global: %d] [Operacion: Suma] -> Resultado: %d\n", blockIdx.x, threadIdx.x, globalIdx, C[globalIdx]);
        }
    }
}

void ejecutarPrueba(int N, int hilosPorBloque, bool verbose, const char* nombrePrueba) {
    int numBlocks = (N + hilosPorBloque - 1) / hilosPorBloque;
    size_t bytes = N * sizeof(int);

    printf("\n=====================================================\n");
    printf("PRUEBA: %s (N = %d, Hilos/bloque = %d)\n", nombrePrueba, N, hilosPorBloque);
    printf("=====================================================\n");

    // Memoria explicita
    int *h_A = new int[N], *h_B = new int[N], *h_C = new int[N];
    int *d_A, *d_B, *d_C;
    cudaMalloc(&d_A, bytes); cudaMalloc(&d_B, bytes); cudaMalloc(&d_C, bytes);
    for (int i = 0; i < N; i++) { h_A[i] = (i % 10) + 1; h_B[i] = (i % 5) + 1; }

    cudaMemcpy(d_A, h_A, bytes, cudaMemcpyHostToDevice);
    cudaMemcpy(d_B, h_B, bytes, cudaMemcpyHostToDevice);

    cudaEvent_t start, stop;
    cudaEventCreate(&start); cudaEventCreate(&stop);

    cudaEventRecord(start);
    sumarArreglos<<<numBlocks, hilosPorBloque>>>(d_A, d_B, d_C, N, verbose);
    cudaDeviceSynchronize();
    cudaEventRecord(stop); cudaEventSynchronize(stop);

    float ms = 0; cudaEventElapsedTime(&ms, start, stop);
    printf("[RESUMEN EXPLICITA] Tiempo de Kernel: %.4f ms\n", ms);

    // Memoria unificada
    int *A, *B, *C;
    cudaMallocManaged(&A, bytes); cudaMallocManaged(&B, bytes); cudaMallocManaged(&C, bytes);
    for (int i = 0; i < N; i++) { A[i] = (i % 10) + 1; B[i] = (i % 5) + 1; }

    cudaEventRecord(start);
    sumarArreglos<<<numBlocks, hilosPorBloque>>>(A, B, C, N, verbose);
    cudaDeviceSynchronize();
    cudaEventRecord(stop); cudaEventSynchronize(stop);

    cudaEventElapsedTime(&ms, start, stop);
    printf("[RESUMEN UNIFICADA] Tiempo de Kernel: %.4f ms\n", ms);

    delete[] h_A; delete[] h_B; delete[] h_C;
    cudaFree(d_A); cudaFree(d_B); cudaFree(d_C);
    cudaFree(A); cudaFree(B); cudaFree(C);
}

int main() {
    printf("=====================================================\n");
    printf("INICIO - PRACTICA 3.2 OPERACIONES ARITMETICAS CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    // PRUEBA 1: 40 elementos (detallada)
    ejecutarPrueba(40, 10, true, "PRUEBA DETALLADA");

    // PRUEBA 2: 100,000 elementos (masiva)
    ejecutarPrueba(100000, 256, false, "PRUEBA MASIVA");

    printf("=====================================================\n");
    printf("FIN - PRACTICA 3.2 OPERACIONES ARITMETICAS CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");
    return 0;
}
```
