```python
%%writefile practica3_5.cu
// =====================================================
// Integrantes:
// [MARIN PUGA ULISES ISRAEL]
// [ZARATE CASTILLO ANDREA ITZEL]
//
// Actividad 3.5: Búsquedas paralelas en CUDA
// =====================================================

#include <cstdio>
#include <cstdlib>
#include <cuda_runtime.h>
#include <ctime>

// =====================================================
// KERNEL: LLENAR ARREGLO CON ALEATORIOS
// =====================================================
__global__ void llenarArreglo(int* arreglo, int N) {
    int idx = blockIdx.x * blockDim.x + threadIdx.x;
    if (idx < N) {
        // Generar aleatorio entre 1 y (2 * N)
        arreglo[idx] = (idx * 17 + 31) % (N * 2) + 1;
    }
}

// =====================================================
// KERNEL: BÚSQUEDA PARALELA LINEAL
// =====================================================
__global__ void buscarValor(int* arreglo, int N, int valorBuscado, bool verbose) {
    int idx = blockIdx.x * blockDim.x + threadIdx.x;
    if (idx < N && arreglo[idx] == valorBuscado) {
        if (verbose) {
            printf("[DEVICE] Bloque: %d, Hilo: %d | Encontrado en la posicion: %d\n", blockIdx.x, threadIdx.x, idx);
        }
    }
}

// =====================================================
// MAIN
// =====================================================
int main() {
    printf("=====================================================\n");
    printf("INICIO - PRACTICA 3.5 BUSQUEDA PARALELA CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    int N = 1000000; // Tamaño para la primera ejecución
    int size = N * sizeof(int);
    int* d_arreglo;
    cudaMalloc(&d_arreglo, size);

    int threadsPerBlock = 32;
    int numBlocks = (N + threadsPerBlock - 1) / threadsPerBlock;

    // Llenar el arreglo en la GPU
    llenarArreglo<<<numBlocks, threadsPerBlock>>>(d_arreglo, N);
    cudaDeviceSynchronize();

    // Copiar a la CPU para imprimirlo (solo en la prueba de 100)
    int* h_arreglo = new int[N];
    cudaMemcpy(h_arreglo, d_arreglo, size, cudaMemcpyDeviceToHost);
    /*
    printf("\nArreglo generado (N=%d):\n[", N);
    for (int i = 0; i < N; i++) {
        printf("%d%s", h_arreglo[i], (i < N - 1) ? ", " : "");
    }
    printf("]\n\n");
    */
    int opcion = 0;
    while (opcion != 2) {
        printf("===== MENU DE BUSQUEDA =====\n");
        printf("1. Buscar un valor\n");
        printf("2. Salir\n");
        printf("Opcion: ");
        scanf("%d", &opcion);

        if (opcion == 1) {
            int valorBuscado;
            printf("Ingrese el valor a buscar: ");
            scanf("%d", &valorBuscado);

            printf("\n--- INICIANDO BUSQUEDA EN GPU ---\n");

            cudaEvent_t start, stop;
            cudaEventCreate(&start); cudaEventCreate(&stop);
            cudaEventRecord(start);

            // Lanzar kernel de busqueda
            buscarValor<<<numBlocks, threadsPerBlock>>>(d_arreglo, N, valorBuscado, true);

            cudaDeviceSynchronize();
            cudaEventRecord(stop); cudaEventSynchronize(stop);

            float ms = 0;
            cudaEventElapsedTime(&ms, start, stop);

            printf("[HOST] Tiempo de busqueda en GPU: %.4f ms\n\n", ms);
        }
    }

    cudaFree(d_arreglo);
    delete[] h_arreglo;

    printf("=====================================================\n");
    printf("FIN - PRACTICA 3.5 BUSQUEDA PARALELA CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    return 0;
}
```
