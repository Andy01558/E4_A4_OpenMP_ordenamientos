```python
%%writefile practica3_6.cu
#include <cstdio>
#include <cstdlib>
#include <cuda_runtime.h>
#include <ctime>

const int TAM_ALFABETO = 36;

// =====================================================
// FUNCIÓN DEVICE: Convertir índice a clave
// =====================================================
__device__ void convertirIndiceAClave(unsigned long long idx, char* clave, int longitud) {
    for (int i = longitud - 1; i >= 0; i--) {
        int val = idx % TAM_ALFABETO;
        if (val < 26) {
            clave[i] = 'A' + val;       // Letras A-Z
        } else {
            clave[i] = '0' + (val - 26); // Números 0-9
        }
        idx /= TAM_ALFABETO;
    }
}

// =====================================================
// KERNEL: FUERZA BRUTA PARALELA
// =====================================================
__global__ void fuerzaBrutaKernel(const char* claveObjetivo, int longitud, unsigned long long rangoInicio, unsigned long long rangoFin, bool* encontrado, unsigned long long* indiceGanador) {
    unsigned long long idGlobal = blockIdx.x * blockDim.x + threadIdx.x;
    unsigned long long totalHilos = gridDim.x * blockDim.x;

    for (unsigned long long idx = rangoInicio + idGlobal; idx <= rangoFin; idx += totalHilos) {
        if (*encontrado) return;

        char intento[11];
        convertirIndiceAClave(idx, intento, longitud);
        intento[longitud] = '\0';

        bool match = true;
        for (int i = 0; i < longitud; i++) {
            if (intento[i] != claveObjetivo[i]) {
                match = false;
                break;
            }
        }

        if (match) {
            if (!(*encontrado)) {
                *encontrado = true;
                *indiceGanador = idx;
                printf("[DEVICE] Bloque: %d, Hilo: %d | CLAVE ENCONTRADA: %s\n", blockIdx.x, threadIdx.x, intento);
            }
            return;
        }
    }
}

// =====================================================
// MAIN
// =====================================================
int main() {
    printf("=====================================================\n");
    printf("INICIO - PRACTICA 3.6 FUERZA BRUTA CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    const char* claveObjetivo = "X7B9K2M4A1";
    int longitud = 10;

    printf("Clave objetivo: %s (Longitud: %d)\n", claveObjetivo, longitud);
    printf("NOTA: El proceso puede tardar unos minutos...\n");

    unsigned long long totalCombinaciones = 1;
    for (int i = 0; i < longitud; i++) totalCombinaciones *= TAM_ALFABETO;
    printf("Total de combinaciones posibles: %llu\n", totalCombinaciones);

    int bloques = 1000;
    int hilosPorBloque = 1024;
    printf("Configuracion: %d bloques x %d hilos = %llu hilos totales.\n", bloques, hilosPorBloque, (unsigned long long)bloques*hilosPorBloque);

    char* d_claveObjetivo;
    bool* d_encontrado;
    unsigned long long* d_indiceGanador;

    cudaMalloc(&d_claveObjetivo, (longitud + 1) * sizeof(char));
    cudaMallocManaged(&d_encontrado, sizeof(bool));
    cudaMallocManaged(&d_indiceGanador, sizeof(unsigned long long));

    *d_encontrado = false;
    *d_indiceGanador = 0;
    cudaMemcpy(d_claveObjetivo, claveObjetivo, (longitud + 1) * sizeof(char), cudaMemcpyHostToDevice);

    cudaEvent_t start, stop;
    cudaEventCreate(&start); cudaEventCreate(&stop);
    cudaEventRecord(start);

    fuerzaBrutaKernel<<<bloques, hilosPorBloque>>>(d_claveObjetivo, longitud, 0, totalCombinaciones - 1, d_encontrado, d_indiceGanador);

    cudaDeviceSynchronize();
    cudaEventRecord(stop); cudaEventSynchronize(stop);

    float ms = 0;
    cudaEventElapsedTime(&ms, start, stop);

    if (*d_encontrado) {
        printf("\n[HOST] CLAVE DESCIFRADA!\n");
        printf("[HOST] Indice ganador: %llu\n", *d_indiceGanador);
    } else {
        printf("\n[HOST] Clave no encontrada en el rango.\n");
    }
    printf("[HOST] Tiempo total de ejecucion: %.4f ms\n", ms);

    cudaFree(d_claveObjetivo);
    cudaFree(d_encontrado);
    cudaFree(d_indiceGanador);
    cudaEventDestroy(start); cudaEventDestroy(stop);

    printf("\n=====================================================\n");
    printf("FIN - PRACTICA 3.6 FUERZA BRUTA CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    return 0;
}
```
