```python
%%writefile practica3_4.cu
// =====================================================
// Integrantes:
// [Apellido1 Apellido2 Nombre1]
// [Apellido1 Apellido2 Nombre2]
// Actividad 3.4: Sincronización en CUDA
// =====================================================

#include <cstdio>
#include <cstdlib>
#include <cuda_runtime.h>
#include <ctime>

// =====================================================
// KERNEL: SIMULACIÓN BANCARIA (100 Hilos, 1 Bloque)
// =====================================================
__global__ void bancoKernel(int* saldo, bool verbose) {
    int tid = threadIdx.x;
    int bid = blockIdx.x;

    // Generamos un numero pseudoaleatorio usando matemática simple (sin rand())
    int semilla = tid + 1;
    int operacion = (semilla % 3) + 1; // 1: Deposito, 2: Retiro, 3: Consulta
    int monto = ((semilla * 37) % 500) + 1;   // Monto entre 1 y 500

    int saldoActual = 0;

    if (operacion == 1) { // Deposito
        atomicAdd(saldo, monto);
        saldoActual = *saldo;
        if (verbose) {
            printf("[DEVICE] Bloque: %d, Hilo: %d | Op: Deposito | Monto: $%d | Nuevo Saldo: $%d\n", bid, tid, monto, saldoActual);
        }
    }
    else if (operacion == 2) { // Retiro
        int saldoTemporal = *saldo;
        if (saldoTemporal >= monto) {
            atomicSub(saldo, monto);
            saldoActual = *saldo;
            if (verbose) {
                printf("[DEVICE] Bloque: %d, Hilo: %d | Op: Retiro | Monto: $%d | Nuevo Saldo: $%d\n", bid, tid, monto, saldoActual);
            }
        } else {
            if (verbose) {
                printf("[DEVICE] Bloque: %d, Hilo: %d | Op: Retiro | Monto: $%d | RECHAZADO (Fondos insuficientes) | Saldo: $%d\n", bid, tid, monto, saldoTemporal);
            }
        }
    }
    else { // Consulta
        saldoActual = *saldo;
        if (verbose) {
            printf("[DEVICE] Bloque: %d, Hilo: %d | Op: Consulta | Monto: $0 | Nuevo Saldo: $%d\n", bid, tid, saldoActual);
        }
    }
}

// =====================================================
// MAIN
// =====================================================
int main() {
    printf("=====================================================\n");
    printf("INICIO - PRACTICA 3.4 Sincronizacion en CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    int* d_saldo;
    cudaMallocManaged(&d_saldo, sizeof(int));
    *d_saldo = 10000;

    printf("Saldo inicial en la cuenta: $%d\n\n", *d_saldo);
    printf("--- EJECUTANDO PRIMERA PRUEBA (100 Hilos, 1 Bloque) ---\n");

    cudaEvent_t start, stop;
    cudaEventCreate(&start); cudaEventCreate(&stop);
    cudaEventRecord(start);

    bancoKernel<<<1, 100>>>(d_saldo, true);

    cudaDeviceSynchronize();
    cudaEventRecord(stop); cudaEventSynchronize(stop);

    float ms = 0;
    cudaEventElapsedTime(&ms, start, stop);

    printf("\n[RESUMEN]\n");
    printf("Saldo final de la cuenta: $%d\n", *d_saldo);
    printf("Tiempo de ejecucion del Kernel: %.4f ms\n", ms);

    cudaFree(d_saldo);
    cudaEventDestroy(start); cudaEventDestroy(stop);

    printf("\n=====================================================\n");
    printf("FIN - PRACTICA 3.4 Sincronizacion en CUDA\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    return 0;
}
```

```python
%%writefile practica3_4_masivo.cu
#include <cstdio>
#include <cstdlib>
#include <cuda_runtime.h>

// Estructura para guardar los resultados de cada hilo
struct ResultadoHilo {
    int bloque;
    int hilo;
    int operacion;
    int monto;
    int saldoFinal;
};

__global__ void bancoKernel(int* saldo, ResultadoHilo* resultados, bool verbose) {
    int tid = threadIdx.x;
    int bid = blockIdx.x;
    int idGlobal = bid * blockDim.x + tid;

    int semilla = idGlobal + 1;
    int operacion = (semilla % 3) + 1; // 1: Deposito, 2: Retiro, 3: Consulta
    int monto = ((semilla * 37) % 500) + 1;

    if (operacion == 1) {
        atomicAdd(saldo, monto);
    }
    else if (operacion == 2) {
        if (*saldo >= monto) {
            atomicSub(saldo, monto);
        }
    }

    // Guardamos los resultados en el arreglo estructurado (solo los primeros 20,000)
    if (verbose && idGlobal < 20000) {
        resultados[idGlobal].bloque = bid;
        resultados[idGlobal].hilo = tid;
        resultados[idGlobal].operacion = operacion;
        resultados[idGlobal].monto = monto;
        resultados[idGlobal].saldoFinal = *saldo;
    }
}

int main() {
    printf("=====================================================\n");
    printf("INICIO - PRACTICA 3.4 MASIVA (1,000,000 Hilos)\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    int* d_saldo;
    cudaMallocManaged(&d_saldo, sizeof(int));
    *d_saldo = 10000;

    // Arreglo en memoria unificada para guardar los resultados
    int maxRegistros = 20000;
    ResultadoHilo* h_resultados;
    cudaMallocManaged(&h_resultados, maxRegistros * sizeof(ResultadoHilo));

    printf("Saldo inicial: $%d\n", *d_saldo);
    printf("Ejecutando 1,000,000 de hilos (1000 bloques x 1000 hilos)...\n");

    cudaEvent_t start, stop;
    cudaEventCreate(&start); cudaEventCreate(&stop);
    cudaEventRecord(start);

    // 1000 bloques x 1000 hilos = 1,000,000 hilos
    bancoKernel<<<1000, 1000>>>(d_saldo, h_resultados, true);

    cudaDeviceSynchronize();
    cudaEventRecord(stop); cudaEventSynchronize(stop);

    float ms = 0;
    cudaEventElapsedTime(&ms, start, stop);

    printf("Saldo final de la cuenta: $%d\n", *d_saldo);
    printf("Tiempo de ejecucion del Kernel: %.4f ms\n", ms);

    // El HOST se encarga de escribir el archivo
    FILE* archivo = fopen("log_banco.txt", "w");
    if (archivo != NULL) {
        fprintf(archivo, "Saldo Final: %d\nTiempo Kernel: %.4f ms\n\n", *d_saldo, ms);
        for (int i = 0; i < maxRegistros; i++) {
            fprintf(archivo, "[DEVICE] Bloque: %d, Hilo: %d | Op: %d | Monto: $%d | Saldo: $%d\n",
                    h_resultados[i].bloque, h_resultados[i].hilo,
                    h_resultados[i].operacion, h_resultados[i].monto,
                    h_resultados[i].saldoFinal);
        }
        fclose(archivo);
        printf("Log guardado exitosamente en 'log_banco.txt' (20,000 registros).\n");
    }

    cudaFree(d_saldo);
    cudaFree(h_resultados);

    printf("=====================================================\n");
    printf("FIN - PRACTICA 3.4\n");
    printf("[MARIN PUGA ULISES ISRAEL]\n");
    printf("[ZARATE CASTILLO ANDREA ITZEL]\n");
    printf("=====================================================\n");

    return 0;
}
```
